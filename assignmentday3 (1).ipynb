```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# Voltage (in Volts)
X = np.array([5, 6, 7, 8, 9, 10]).reshape(-1, 1)

# Power consumption (in Watts)
y = np.array([10, 12, 14, 16, 18, 20])

# Create Linear Regression model
model = LinearRegression()

# Train the model
model.fit(X, y)

# Predict power consumption
y_pred = model.predict(X)

# Print model parameters
print("Slope:", model.coef_[0])
print("Intercept:", model.intercept_)

# Predict power for 12 Volts
voltage = np.array([[12]])
predicted_power = model.predict(voltage)

print("Predicted power consumption at 12V:",
      predicted_power[0], "Watts")

# Plot the data and regression line
plt.scatter(X, y, label="Actual Data")
plt.plot(X, y_pred, label="Regression Line")

plt.xlabel("Voltage (V)")
plt.ylabel("Power Consumption (W)")
plt.title("Linear Regression - Electronics Industry")
plt.legend()
plt.show()
```

. Demonstrate application of unsupervised learning to train a model

```python

import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

# Create sample dataset
data = {
    'Annual_Income': [15, 16, 17, 18, 20, 22,
                      50, 52, 55, 58, 60, 62,
                      85, 88, 90, 92, 95, 98],

    'Spending_Score': [80, 85, 78, 82, 88, 75,
                       40, 45, 42, 38, 35, 40,
                       80, 85, 88, 82, 90, 87]
}

df = pd.DataFrame(data)

print(df)
```

Apply K-Means Clustering

```python
# Select features
X = df[['Annual_Income', 'Spending_Score']]

# Create K-Means model
model = KMeans(n_clusters=3, random_state=42, n_init=10)

# Train the model
model.fit(X)

# Get cluster labels
df['Cluster'] = model.labels_

print("\nClustered Data:")
print(df)
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(
    df['Annual_Income'],
    df['Spending_Score'],
    c=df['Cluster'],
    s=100
)

# Plot cluster centers
centers = model.cluster_centers_

plt.scatter(
    centers[:, 0],
    centers[:, 1],
    marker='X',
    s=200
)

plt.xlabel("Annual Income")
plt.ylabel("Spending Score")
plt.title("Customer Segmentation using K-Means")
plt.show()
```
